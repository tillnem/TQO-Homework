```python
import numpy as np
import scipy.linalg
import scipy.sparse as sp
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
```

```python
from scipy.optimize import fsolve
import matplotlib.pyplot as plt

A = 1

def equation(x, D, U):
    return x - A**2 / ((D + U*x)**2 + 1)

D_values = np.linspace(-5, 5, 200)
x_solutions1 = []
x_solutions2 = []
x_solutions3 = []
x_solutions4 = []
# For each D, solve for x
for D in D_values:
    # Provide an initial guess, e.g., x0 = 0
    sol1 = fsolve(equation, x0=0, args=(D, 0.1))
    x_solutions1.append(sol1[0])

for D in D_values:
    # Provide an initial guess, e.g., x0 = 0
    sol2 = fsolve(equation, x0=0, args=(D,0.5))
    x_solutions2.append(sol2[0])

for D in D_values:
    # Provide an initial guess, e.g., x0 = 0
    sol3 = fsolve(equation, x0=0, args=(D,1.62))
    x_solutions3.append(sol3[0])

for D in D_values:
    # Provide an initial guess, e.g., x0 = 0
    sol4 = fsolve(equation, x0=0, args=(D,1.8))
    x_solutions4.append(sol4[0])

plt.plot(D_values, x_solutions1, label='U = 0.1')
plt.plot(D_values, x_solutions2, label='U = 0.5')
plt.plot(D_values, x_solutions3, label='U= 1.62')
plt.plot(D_values, x_solutions4, label='U = 1.8')
plt.xlabel('Delta')
plt.ylabel('x (solution)')
plt.title('Observation of optical bistability')
plt.legend()
plt.show()
```

We can see here very well that from U~1.6 upwards, we get the instability effects where the solutions aren't really continuous anymore and we can observe the bistability
