```python
import numpy as np
import scipy.linalg
import scipy.sparse as sp
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 3, 100)
y1 = x
D = -2
A = 1
U = 1.7
y2 = A**2/((D+U*x)**2+1)
plt.plot(x, y1, label='n')
plt.plot(x, y2, label='E=1, U = 1.7, D=-2   ')
plt.xlabel('n')
plt.title('Graphical discussion')
plt.legend()
```
