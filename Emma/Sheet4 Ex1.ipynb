```python
import numpy as np
import scipy.linalg
import scipy.sparse as sp
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
import time
np.set_printoptions(precision=3)
```

```python
def funktion(t, lamb, omega_m):
    return 0.5*(1+np.cos((lamb**2/omega_m)*(t+np.sin(omega_m*t)))*np.exp(-((lamb/(2*omega_m))**2)*np.abs(1-np.exp(-1j*omega_m*t))))
t = np.linspace(0,100,200)
plt.plot(t,funktion(t, 1,0.5), label = r"$\lambda$ = 1, $\omega_m$=0.5 ")
plt.plot(t,funktion(t, 1,1), label = r"$\lambda$ = 1, $\omega_m$=1 ")
plt.plot(t,funktion(t, 1,2), label = r"$\lambda$ = 1, $\omega_m$=2 ")
#plt.plot(t,funktion(t, 1,2), label = r"$\lambda$ = 1, $\omega_m$=3 ")
plt.xlabel("t")
plt.legend()
plt.title("Time evolution of the excited state")
plt.ylabel("Excited state probability")
```

We can see that we get revivals in the Probability. For big omega, the population drops to zero eventually. This does not happen for smaller omegas.
